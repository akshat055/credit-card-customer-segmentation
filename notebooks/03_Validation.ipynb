```python
import pandas as pd
from pathlib import Path

# Locate data folder
data_path = Path("../data")

# Load datasets
original = pd.read_csv(data_path / "clustered_customers.csv")
predicted = pd.read_csv(data_path / "customer_predictions.csv")

print("Original dataset:", original.shape)
print("Predicted dataset:", predicted.shape)

print("\nOriginal columns:")
print(original.columns.tolist())

print("\nPredicted columns:")
print(predicted.columns.tolist())
```

```python
from sklearn.metrics import (
    accuracy_score,
    adjusted_rand_score,
    confusion_matrix,
    classification_report
)

# Original and predicted cluster labels
y_original = original["Cluster"]
y_predicted = predicted["Predicted Cluster"]

# Compare labels
accuracy = accuracy_score(y_original, y_predicted)
ari = adjusted_rand_score(y_original, y_predicted)

print("Exact Cluster Agreement:", round(accuracy * 100, 2), "%")
print("Adjusted Rand Index:", round(ari, 4))

# Confusion matrix
print("\nConfusion Matrix:")
print(confusion_matrix(y_original, y_predicted))

# Cluster-wise comparison
comparison = pd.crosstab(
    y_original,
    y_predicted,
    rownames=["Original Cluster"],
    colnames=["Predicted Cluster"]
)

display(comparison)
```
